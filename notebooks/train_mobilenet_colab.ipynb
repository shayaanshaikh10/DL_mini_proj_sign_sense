## 8. Bring the results back to your laptop

The next cell prints the trained model and results as one long block of text.
After it finishes: **save this notebook (Ctrl+S)**, then on your laptop run

`python scripts/import_colab_outputs.py`

It reads that text from the saved notebook and recreates `models/mobilenetv2.keras` and the `results/*mobilenetv2*` files.

```python
# 8. Print the outputs as text (do not clear this output before saving!)
import glob
buffer = io.BytesIO()
with zipfile.ZipFile(buffer, 'w', zipfile.ZIP_DEFLATED) as archive:
    for path in glob.glob('models/mobilenetv2*.keras') + glob.glob('results/*mobilenetv2*'):
        archive.write(path)
print('SIGNSENSE_OUTPUTS_START')
print(base64.b64encode(buffer.getvalue()).decode('ascii'))
print('SIGNSENSE_OUTPUTS_END')
```
